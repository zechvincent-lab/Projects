# **Feature Engineering**

```python
#Ensure dates are datetime
data['Delivery_Date'] = pd.to_datetime(data['Delivery_Date'])
data['Customer_BirthDate'] = pd.to_datetime(data['Customer_BirthDate'])

#Currency to float
def currency_to_float(value):
    return float(value.replace('$', '').replace(',', ''))

#Frequency
data['Frequency'] = data.groupby('Customer ID')['Order ID'].transform('count')

#Recency
data['Recency'] = (pd.Timestamp('today') - data['Delivery_Date']).dt.days

#CLV
data['CLV'] = data['Total Revenue'].apply(currency_to_float)

#Average Unit Cost
data['Avg_Unit_Cost'] = data['Unit Cost'].apply(currency_to_float)

#Customer Age
today = pd.Timestamp('today')
data['Age'] = (today - data['Customer_BirthDate']).dt.days // 365
```

```python
#One Customer Per Row
agg_operations = {
    'Frequency': 'max',
    'Recency': 'min',
    'CLV': 'sum',
    'Avg_Unit_Cost': 'mean',
    'Age': 'first'
}

data_aggr = data.groupby('Customer ID').agg(agg_operations).reset_index()

#Reduce the amount of features
data_aggr = data_aggr[['Customer ID', 'Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']]

data_aggr.head()
```

# **Outliers**

Looking at our boxplots, we can see many outliers across each of our new features. A quick look using the IQR (Interquartile Range) method can highlight the customer IDs with the most anomalies to shed light on their trends and behaviours, helping us determine whether they are true anomalies or errors.

```python
#IQR

q1 = data_aggr.quantile(0.25)
q3 = data_aggr.quantile(0.75)
iqr = q3 - q1

#limits
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

#adding the anomalies
data_aggr_anomalies = data_aggr.copy()

for col in data_aggr_anomalies.columns:
    data_aggr_anomalies[col + '_anomaly'] = (((data_aggr_anomalies[col] < lower_limit[col]) | (data_aggr_anomalies[col] > upper_limit[col])).astype(int))

#adding outlier total colomn

anomaly_cols = [col for col in data_aggr_anomalies.columns if '_anomaly' in col]
data_aggr_anomalies['Outlier_Total'] = data_aggr_anomalies[anomaly_cols].sum(axis=1)

#combination of anomalies

total_rows = len(data_aggr_anomalies)
total_anomalies = data_aggr_anomalies.groupby('Outlier_Total').size().reset_index(name='Count')
total_anomalies['percentage'] = (total_anomalies['Count'] / total_rows) * 100

print(total_anomalies)
```

#**Feature Scaling**





```python
#Features for Scaling
features_for_clustering = ['Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']
```

```python
data_aggr = data_aggr.reset_index(drop=True)
```

```python
data_features = data_aggr[features_for_clustering]
```

```python
#Pipeline for preprocessing

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), features_for_clustering)
    ])
```

```python
#Fit and Transform the data

data_preprocessed = preprocessor.fit_transform(data_features)
```

# **Clustering**

# **Elbow Method**

```python
#Elbow Method
wcss = []
for i in range(1, 15):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init='auto')
    kmeans.fit(data_preprocessed)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 15), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

**Elbow Method - conclusion**

Inspection of the elbow method indicates that the reduction in inertia starts to diminish around 5–6 clusters. However, the elbow is not clearly defined, and alternative interpretations could justify a larger number of clusters.

# **Silhouette Method**

```python
#Silhouette Method

for n_clusters in range(2, 8):

    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)

    ax1.set_xlim([-0.1, 1])

    ax1.set_ylim([0, len(data_aggr) + (n_clusters + 1) * 10])

    clusterer = KMeans(n_clusters=n_clusters, random_state=42, n_init='auto')
    cluster_labels = clusterer.fit_predict(data_preprocessed)

    silhouette_avg = silhouette_score(data_preprocessed, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    sample_silhouette_values = silhouette_samples(data_preprocessed, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):

        ith_cluster_silhouette_values = sample_silhouette_values[cluster_labels == i]

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = plt.cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        y_lower = y_upper + 10

    ax1.set_title("The silhouette plot for the various clusters.")
    ax1.set_xlabel("The silhouette coefficient values")
    ax1.set_ylabel("Cluster label")

    ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

    ax1.set_yticks([])
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

    pca = PCA(n_components=2)
    data_2d = pca.fit_transform(data_preprocessed)
    centers_2d = pca.transform(clusterer.cluster_centers_)

    colors = plt.cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)

    ax2.scatter(data_2d[:, 0], data_2d[:, 1], marker='.', s=30, lw=0, alpha=0.7,
              c=colors, edgecolor='k')

    ax2.scatter(centers_2d[:, 0], centers_2d[:, 1], marker='o',
              c="white", alpha=1, s=200, edgecolor='k')

    ax2.set_title("The visualization of the clustered data. (PCA)")
    ax2.set_xlabel("PCA 1")
    ax2.set_ylabel("PCA 2")

    plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                "with n_clusters = %d" % n_clusters),
                fontsize=14, fontweight='bold')

    plt.show()
```

**Silhouette Score - Results**

Evaluating results across multiple K-means cluster counts to identify an optimal value. The silhouette scores indicate generally soft clustering across all configurations. Similar to the elbow method, the results are not definitive, with some clusters appearing to partially overlap or merge. However, solutions with 5–6 clusters appear to offer the most reasonable balance.

# **Hierarchical Method**

```python
#Sample dataset in order to perform hierarchical method due to computerial contratiants

data_sample = data_aggr.sample(n=30000, random_state=42).reset_index(drop=True)

data_sample.head()
```

```python
#sample features and scale
sample_features = data_sample[features_for_clustering]

scaler = StandardScaler()
sample_scaled = scaler.fit_transform(sample_features)
```

```python
#Hierarchical Method
Z = linkage(sample_scaled, method='ward')

dendrogram(Z, truncate_mode='lastp', p=25, leaf_rotation=90, leaf_font_size=12)

plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.show()
```

**Hierarchical Methods - Results**

The hierarchical clustering method is computationally intensive, so a sample of 30,000 records was used. Inspection of the dendrogram shows a relatively short linkage distance between the red and green groups at three clusters, suggesting limited separation at this level. At lower cut points, the linkage distances increase substantially, indicating clearer separation. Based on this, a five-cluster solution appears to be the most appropriate.

# **Optimum K**

After evaluating all three methods, the optimal number of clusters appears to lie in the range of 5–6, although none provide a definitive result. Given that the hierarchical method provides the clearest separation at five clusters, this value will be used as the basis for subsequent analysis.

```python
#initialize the kmeans algorithm
kmeans = KMeans(n_clusters=5, random_state=42)

#create a pipeline that first preprocesses the data and then applies k-means
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('Clusterer', kmeans)
])

#fit the pipleline to the data
pipeline.fit(data_aggr)

#cluster labels
cluster_labels = pipeline.predict(data_aggr)

#add cluster labels to dataframe
data_aggr['Cluster'] = cluster_labels

#add the cluster labels to the DataFrame
data_aggr['Cluster'] = cluster_labels

data_aggr.head()
```
